```python
# Definicion del catalogo y esquema

CATALOGO = "workspace"
ESQUEMA = "ingenieria_datos_miguelzuleta"
```

```python

spark.sql(f"CREATE SCHEMA IF NOT EXISTS {CATALOGO}.{ESQUEMA}")
```

```python
from pyspark.sql.types import StructType, StructField, IntegerType, StringType, DoubleType, DateType
from datetime import datetime

# Definicion explicita del esquema de datos
schema = StructType([
    StructField("id_venta", IntegerType(), True),
    StructField("fecha", StringType(), True),
    StructField("producto", StringType(), True),
    StructField("categoria", StringType(), True),
    StructField("cantidad", IntegerType(), True),
    StructField("precio_unitario", DoubleType(), True)
])

# Conjunto de 16 registros de prueba
# Incluye intencionalmente:
# - Registro 1 duplicado de forma identica
# - Registro 14 sin producto (None)
# - Registro 15 con cantidad negativa (-2)
data = [
    (1, "2026-03-01", "Teclado Mecanico", "Electronica", 2, 45.0),
    (2, "2026-03-01", "Raton Inalambrico", "Electronica", 1, 25.0),
    (3, "2026-03-02", "Monitor 24 Pulgadas", "Electronica", 1, 150.0),
    (4, "2026-03-02", "Camiseta Algodon", "Ropa", 3, 15.0),
    (5, "2026-03-03", "Pantalon Jean", "Ropa", 2, 40.0),
    (6, "2026-03-03", "Chaqueta Impermeable", "Ropa", 1, 80.0),
    (7, "2026-03-04", "Cafetera Goteo", "Hogar", 1, 35.0),
    (8, "2026-03-04", "Licuadora 500W", "Hogar", 2, 50.0),
    (9, "2026-03-05", "Juego de Sabanas", "Hogar", 1, 30.0),
    (10, "2026-03-05", "Audifonos Bluetooth", "Electronica", 2, 60.0),
    (11, "2026-03-06", "Zapatos Deportivos", "Ropa", 1, 75.0),
    (12, "2026-03-06", "Lampara Escritorio", "Hogar", 3, 20.0),
    (13, "2026-03-07", "Mesa Auxiliar", "Hogar", 1, 45.0),
    (14, "2026-03-07", None, "Electronica", 1, 99.0),
    (15, "2026-03-08", "Cargador Rapido USB-C", "Electronica", -2, 18.0),
    (1, "2026-03-01", "Teclado Mecanico", "Electronica", 2, 45.0)
]

# Construccion del DataFrame
df_ventas = spark.createDataFrame(data=data, schema=schema)
```

```python
# Lo guardamos en la tabla Delta ventas_bronze
tabla_destino = f"{CATALOGO}.{ESQUEMA}.ventas_bronze"

df_ventas.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable(tabla_destino)
```
