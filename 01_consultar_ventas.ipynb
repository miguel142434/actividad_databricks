```python
# Definicion del catalogo y esquema
CATALOGO = "workspace"
ESQUEMA = "ingenieria_datos_miguelzuleta"
TABLA = f"{CATALOGO}.{ESQUEMA}.ventas_bronze"
```

```python
# Lectura de la tabla Delta usando su nombre completo
df_leido = spark.read.table(TABLA)
```

```python
# Consulta del total de registros
total_registros = df_leido.count()
print(f"Total de registros en la tabla {TABLA}: {total_registros}")
```

```python
# Filtrado y visualizacion de ventas de una categoria especifica
categoria_seleccionada = "Electronica"

df_categoria = df_leido.filter(df_leido.categoria == categoria_seleccionada)
display(df_categoria)
```
